# Analyst comparison template

This notebook uses the supported JapanTrade analytics API for country rankings, product-country comparisons, and coverage checks.

```python
from pathlib import Path

from japantrade import (
    country_product_ranking,
    coverage_report,
    load_normalized_data,
    product_country_comparison,
    search_hs,
)

DATA_PATH = Path('data/japan_exports_2023_2025.parquet')
df = load_normalized_data(DATA_PATH)
df[['direction', 'date']].agg(['min', 'max'])
```

```python
# Search before selecting an unfamiliar product code.
search_hs('electric vehicle', level=4)
```

```python
# The default compares the latest 12 available months with the previous 12.
italy_ranking = country_product_ranking(
    df, country='Italy', direction='export', hs_level=4, limit=20
)
italy_ranking
```

```python
comparison = product_country_comparison(
    df, countries=['Italy', 'Germany', 'United States'], codes=['8703'], direction='export'
)
comparison
```

```python
# Check coverage if a comparison window is incomplete.
coverage_report(df[df['direction'] == 'export']).head()
```
